```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
```

```python
G = nx.Graph()

for i in range(4):
    G.add_node(i)

for i in range(5):
    edge_to = np.random.randint(4)
    edge_from = np.random.randint(4)
    if edge_to != edge_from:
        G.add_edge(edge_to, edge_from)
```

```python
fig = plt.figure(figsize=(6,6))

nx.draw_networkx(G)
plt.title("Sample Network Graph")
plt.savefig("output_files/4_node_network.png", dpi=400)
```

```python
D = nx.DiGraph()

for i in range(4):
    D.add_node(i)

for i in range(5):
    edge_to = np.random.randint(4)
    edge_from = np.random.randint(4)
    if edge_to != edge_from:
        D.add_edge(edge_to, edge_from)
```

```python
fig = plt.figure(figsize=(6,6))

nx.draw_networkx(D)
plt.title("Sample Directed Network Graph")
plt.savefig("output_files/4_node_directed_network.png", dpi=400)
```
