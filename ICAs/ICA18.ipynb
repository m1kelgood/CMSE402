```python
import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd
import plotly.express as px
```

```python
df = pd.read_excel("data_files/EDGAR_AR5_GHG_1970_2023.xlsx", sheet_name="TOTALS BY COUNTRY", skiprows=9)
df.head(5)
```

```python
fig = px.choropleth(
    df,
    locations="Country_code_A3",
    color="Y_1970",
    projection="natural earth",
    title="Green House Gas Emissions in 1970 by Country",
    hover_data=["Name", "Y_1970"],
    color_continuous_scale="Reds",
    
)

fig.write_html("output_files/GHG_1970_choro.html", include_plotlyjs="cdn")
fig.show()
```

```python
fig = px.choropleth(
    df,
    locations="Country_code_A3",
    color="Y_2023",
    projection="natural earth",
    title="Green House Gas Emissions in 2023 by Country",
    hover_data=["Name", "Y_2023"],
    color_continuous_scale="Reds",
    
)

fig.write_html("output_files/GHG_2023_choro.html", include_plotlyjs="cdn")
fig.show()
```
