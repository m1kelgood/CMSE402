```python
import PIL
from PIL import Image
from PIL import ImageFilter
from scipy import ndimage
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = Image.open("data_files/test_image.png")
```

```python
plt.imshow(img)
plt.show()
```

```python
print("Image format: {}, image size: {}, image mode: {}.".format(img.format, img.size, img.mode))
```

```python
crop = img.crop((600, 200, 1000, 600))
plt.imshow(crop)
plt.axis("off")
plt.show()
```

```python
crop_90r = crop.rotate(90)
plt.imshow(crop_90r)
```

```python
img_a = img.convert("RGBA")
crop_90ra = crop_90r.convert("RGBA")
img_a.paste(crop_90ra, (600, 200), crop_90ra)
plt.imshow(img_a)
```

```python
flipped = img.transpose(Image.FLIP_TOP_BOTTOM)
flipped_r = flipped.rotate(45)
plt.imshow(flipped_r)
```

```python
gray = img.convert("L")
gray
```

```python
contour = img.filter(ImageFilter.CONTOUR)
plt.imshow(contour)
plt.axis("off")
plt.show()
```

```python
edge = img.filter(ImageFilter.EDGE_ENHANCE)
plt.imshow(edge)
plt.axis("off")
plt.show()
```

```python
sharp = img.filter(ImageFilter.SHARPEN)
plt.imshow(sharp)
plt.axis("off")
plt.show()
```

```python
im_ary = np.array(img)
im_ary.shape
```

```python
red_only = im_ary.copy()
green_only = im_ary.copy()
blue_only = im_ary.copy()

red_only[:, :, 1:] = 0
green_only[:, :, 0] = 0
green_only[:, :, 2] = 0
blue_only[:, :, 0:2] = 0
```

```python
plt.imshow(red_only)
plt.axis("off")
plt.show()
```

```python
plt.imshow(green_only)
plt.axis("off")
plt.show()
```

```python
plt.imshow(blue_only)
plt.axis("off")
plt.show()
```

```python
ch1 = im_ary[:, :, 0]
ch2 = im_ary[:, :, 1]

swap = im_ary.copy()
swap[:, :, 0] = ch2
swap[:, :, 1] = ch1

plt.imshow(swap)
plt.axis("off")
plt.show()
```
