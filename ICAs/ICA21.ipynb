```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
hartford_web = nx.read_edgelist("data_files/hartford_drug.txt", nodetype=int, create_using=nx.DiGraph)
pos = nx.forceatlas2_layout(hartford_web)
```

```python
fig = plt.figure(figsize=(12,8))

nx.draw_networkx_nodes(
    hartford_web,
    pos=pos,
    node_color="lightgreen",
    alpha=0.7,
    edgecolors="black",
    node_size=300
)

nx.draw_networkx_edges(
    hartford_web,
    pos=pos,
    edge_color="gray",
    width=1.5,
    style="--",
    arrows=True
)


nx.draw_networkx_labels(
    hartford_web,
    pos=pos,
    font_size=6,
    font_color="black"
)

plt.title("Hartford Drug Ring Grouping", fontweight="bold")
plt.axis("off")
plt.show()
```

```python
num_nodes = hartford_web.number_of_nodes()
num_edges = hartford_web.number_of_edges()
connectivity = num_edges / num_nodes

print("Number of nodes: {}, edges: {}, connectivity: {}".format(num_nodes, num_edges, connectivity))
```

As the connectivity of the graph increases, the degree (ratio of edges to nodes) should also increase

```python
out_vals = dict(hartford_web.out_degree).values()
in_vals = dict(hartford_web.in_degree).values()
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 10))

axes[0].hist(out_vals, bins=6, color="royalblue")
axes[0].set_title("Edge Degree Distribution (out)", fontweight="bold")
axes[0].grid(alpha=0.2)

axes[1].hist(in_vals, bins=8, color="seagreen")
axes[1].set_title("Edge Degree Distribution (in)", fontweight="bold")
axes[1].grid(alpha=0.2)

plt.xlabel("Number of Edges")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
degree = dict(hartford_web.degree)
max_node = max(degree, key=degree.get)
min_node = min(degree, key=degree.get)

pos_max = pos[max_node]
pos_min = pos[min_node]
```
